## Задание 10. Функция: расчёт среднего

Создайте функцию `mean_score(scores)`, которая возвращает среднее арифметическое списка. Для пустого списка возвращайте `None`. Проверьте `[60, 90]` и `[]`.

**Подсказка :** Сначала обработайте пустой список, иначе будет деление на ноль.

```python
def mean_score(scores):
    if len(scores) == 0:
        return None
    return sum(scores) / len(scores)

print(mean_score([60, 90]))
print(mean_score([]))
```

## Задание 12. Исключения: ошибочный ввод

Есть список `raw = ["91", "ошибка", "72", "", "84"]`. Пройдитесь по нему и превратите корректные значения в числа. Для остальных выведите предупреждение `Пропуск: ...`. В конце напечатайте список `[91, 72, 84]`.

**Подсказка :** `int()` может вызвать `ValueError`. Ловите именно это исключение.

```python
raw = ["91", "ошибка", "72", "", "84"]

numbers = []
for value in raw:
    try:
        numbers.append(int(value))
    except ValueError:
        print(f"Пропуск: {value}")

print(numbers)
```

## Задание 13. Классы: модель абитуриента

Опишите класс `Applicant` с `__init__(self, name, scores)` и методом `total(self)`, возвращающим сумму баллов. Создайте объект для Анны с баллами `[80, 90, 75]` и выведите имя и сумму.

**Подсказка :** `self` — ссылка на текущий объект; метод вызывается через точку.

```python
class Applicant:
    def __init__(self, name, scores):
        self.name = name
        self.scores = scores

    def total(self):
        return sum(self.scores)

anna = Applicant("Анна", [80, 90, 75])
print(anna.name, anna.total())
```

## Задание 14. Итоговая задача: анализ поступающих

Дан набор из пяти поступающих (код ниже). Напишите функцию `analyze(students, threshold=230)`, возвращающую словарь с ключами `count`, `mean_total`, `passed_names`. `passed_names` — имена тех, чья сумма баллов **не меньше** порога. Для пустого списка `mean_total` должно быть `None`. Выведите результат.

**Подсказка :** Сначала посчитайте суммы, затем отберите имена. Не забудьте обработать пустой список.

```python
students = [
    {"name": "Анна", "scores": [85, 90, 88]},
    {"name": "Борис", "scores": [60, 68, 70]},
    {"name": "Вера", "scores": [79, 82, 84]},
    {"name": "Глеб", "scores": [70, 75, 80]},
    {"name": "Даша", "scores": [90, 96, 95]},
]
```

```python
def analyze(students, threshold=230):
    if len(students) == 0:
        return {"count": 0, "mean_total": None, "passed_names": []}

    totals = [sum(s["scores"]) for s in students]
    passed_names = [s["name"] for s in students if sum(s["scores"]) >= threshold]

    return {
        "count": len(students),
        "mean_total": sum(totals) / len(totals),
        "passed_names": passed_names,
    }

print(analyze(students))
```
